# Valores ausentes

Usualmente, valores ausentes em Python são representados por `None` e `np.nan`.  

```python
val = None
type(val)
```

```python
import numpy as np

val = np.nan

# nan é do tipo float, definido no padrão IEEE 754 de representação de números de ponto flutuante
type(val)
```

```python
# Segundo o padrão, nan não é igual a nenhum outro valor, inclusive a ele mesmo
np.nan == np.nan
```

```python
# Qualquer operação com nan resulta em nan
2*np.nan
```

```python
# Uma série/coluna de inteiros não pode conter valores ausentes, portanto, o Pandas converte a série para float
s = pd.Series([2, 3, 4, np.nan])
s
```

```python
# Em uma série de strings, valores ausentes podem ser representados por None ou np.nan
s = pd.Series(["a", "b", "c", None, np.nan])
s
```

```python
# Uma série de booleanos não pode conter valores ausentes, portanto, o Pandas converte a série para o tipo object
s1 = pd.Series([True, False])
s2 = pd.Series([True, False, np.nan])
print(s1.dtype, s2.dtype)
```

```python
# Perigo! np.nan é considerado True em Python! Isso pode esconder valores ausentes.
s = pd.Series([True, False, np.nan], dtype=bool)
s
```

Os tipos base do Pandas são `int`, `float`, `bool` e `str`.

O Pandas possui os tipos especiais `Int64`, `Float64`, `boolean` e `string`, que permitem a representação de valores ausentes com o tipo `pd.NA`.

```python
s1 = pd.Series([2, 3, 4, np.nan], dtype="Int64")    
s2 = pd.Series(["a", "b", "c", None, np.nan], dtype="string")
s3 = pd.Series([True, False, np.nan], dtype=bool)

display(s1, s2, s3)
```

O leitor CSV do Pandas possui o parâmetro `na_values`, que permite a especificação de valores que devem ser interpretados como ausentes:

```python
df = pd.read_csv(file, na_values=["?"])
df.info()

```
